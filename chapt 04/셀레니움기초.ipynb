#### 웹 페이지 자동 수집

### 셀레니움

- 웹 브라우저를 자동으로 제어하는 파이썬 라이브러리 (패키지)
- 기능
 
  - 웹페에지 로드 (접속)
  - 입력창에 글자 입력 (아이디 패스워드 등)
  - 버튼 클릭
  - 페이지 이동
  - 화면에 표시 된 텍스트 추출
  - 특정 이미지 저장/ 화면 스크린 캡처
  - 반복 자동화

- 정적인HTML은 Beutifulsoup4, request패키지로 수집가능
- 동적인 웹화면은 selenium이 강력함. 정적인 웹페에지도 가능

#### 요소찾기

- 주요 요소 찾기 방식
  - 'By ID': ID로 찾기
  - 'By CLASS_NAME': class 명으로 찾기
  - 'By TAG_NAME':태그명으로 찾기
  - 'By CSS_SELECTOR': CSS 선택자 방법으로 찾기
  -''By XPATH':Xpath로 찾기 
  

```python
#css 선택자로 요소찾는 함수용 패키지 내 클래스 로드
from selenium.webdriver.common.by import By
```

```python
# 페이지 내 하나의 데이터만 송출

title = driver.find_element(By.ID,'title')
title.text
```

#### 명시적 대기

- 링크 클릭으로 다른 페이지 이동, 검색 후 결과 대기까지 시간 소요
- time,sleep(2) '2초동안 대기'등의 파이썬 기본 함수가 제약 사항이 많음
- 특정 조건이 만족할때까지 대기하는 로직 사용

```python
#명시 대기를 위한 패키지 로드

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
#여러 작업 수행 후

wait = WebDriverWait(driver, 10)

title = wait.until(
    EC.presence_of_element_located((By.ID, 'title'))
)

title.text
```

#### 정리

-웹 스크래핑시 필요요소
  웹드라이버로 웹사이트 오픈
  - 필요 요소 찾기
  - 입력태그 찾기
  - 버튼 요소 찾기
  - 입력하고 버튼 클릭해서 동작
  - 데이터 가져오기
  - 화면 갱신될 때까지 명시적 대기

#### 데이터 수집하기

```python
driver.get("http://127.0.0.1:5500/chapt%2004/products.html")

products = driver.find_elements(By.CLASS_NAME, "product")
result = []

for product in products:
    name = product.find_element(By.CLASS_NAME, "name").text
    price = product.find_element(By.CLASS_NAME, "price").text
    result.append({"name": name, "price": price})
```

```python
result 
```

```python
df = pd.DataFrame(result)
df
```

```python
df.to_csv ('../data/scrap_products.csv',encoding ='utf-8')
```

```python
songs= driver.find_elements(By.CSS_SELECTOR,'.lst50,.lst100')
#songs= driver.find_elements(By.CSS_SELECTOR,'tr.lst50, tr.lst100')

datas =[]
#len(songs)

for song in songs:

    #노래 순위
    rank = song.find_element(By.CSS_SELECTOR,'.rank').text
    #노래제목
    title = song.find_element(By.CSS_SELECTOR,'.rank01>span>a').text
    #가수
    singer = song.find_element(By.CSS_SELECTOR,'.rank02>a').text
    #앨범
    album = song.find_element(By.CSS_SELECTOR,'.rank03>a').text
    #좋아요 개수
    like = song.find_element(By.CLASS_NAME,'cnt').text
 
    #단계상승하락
    rank_ud = song.find_element(By.CSS_SELECTOR, ".rank_wrap").get_attribute("title")
   

    datas.append({
        '순위':rank,
        '노래제목':title,
        '가수':singer,
        '앨범':album,
        '좋아요':like,
        '단계상승하락':rank_ud


    })
    
```
