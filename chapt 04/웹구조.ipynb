#### 웹 페이지

-HTML, CSS, Javascrpit로 구성
  - HTML: 웹 페이지 구성, 구조, 내용
  - CSS: 글자, 색상, 배치 등의 디자인 파일
  - Javascript: 클릭, 입력, 화면 변경, 애니메이션 등 동작처리

4. DataFrame 저장/ 이미지파일로 저장

#### HTML 주요태그
| 태그 | 의미 |
| --- | --- |
| `<html>` | HTML 문서 전체 |
| `<head>` | 문서 정보 |
| `<body>` | 페이지 화면에 보이는 내용 |
| `<title>` | 웹브라우저 탭 제목 |
| `<h1>~<h6>` | 제목 |
| `<p>` | 문단 |
| `<a>` | 링크/페이지 이동 |
| `<img>` | 이미지 |
| `<ul>`, `<li>` | 목록, 번호목록 |
| `<table>` | 표, 테이블 |
| `<input>` | 텍스트 입력창, 콤보박스 |
| `<button>` | 버튼 |
| `<div>`, `<span>`, `<section>`, `<content>` | 영역 구분 태그 |

##### 주요 속성

HTML 태그에는 여러 속성이 붙을 수 있음
```html
<div class="product" id="item-1">

    <span class="name">키보드</span>

    <span class="price">20000</span>

</div>




|속성|의미|
|---|---|
|`id`| 웹페이지 내 중복되지 않은 값
|`class`|css디자인을 위해서 사용 여러 요소에 공통으로 붙이는 이름 , 중복됨 |
|`href'|링크주소|
|`src`|이미지나 소스 위치|
|`name'|입력요소 이름 , 중복됨

#### 개발자 도구

웹 페이지에서 원하는 요소를 찾으려면 브라우저 개발자 도구 사용 F12로 동작 

- 웹 페이지 우클릭> 검사
- 요소 탭에서 필요 요소 찾기
- HTML 구조 확인
- 스크래핑 할 데이터 위치 찾기
- 태그 이름, ID, class 확인

##### 선택자

- 태그명 :h1,span,ul...등 기본 태그 선택
- .클래스명: 태그내 class 속성을 선택
- ##아이디명: 태그내 id 속성 값 선택
- .하위태그: 



#### 웹 스크래핑을 위한 구조 이해

- html구조 파악
- 태그에 있는 id와 class 위치 확인
- css파일이나 js파일 자체는 분석 필요없음
