### 기본데이터

```python
data = {

    "order_date": [

        "2026-09-01", "2026-09-01", "2026-09-02", "2026-09-02",

        "2026-09-03", "2026-09-03", "2026-09-04", "2026-09-04",

        "2026-09-05", "2026-09-05"

    ],

    "customer": ["홍길동", "김철수", "이영희", "홍길동", "박민수", "김철수", "최지훈", "이영희", "홍길동", "박민수"],

    "category": ["주변기기", "주변기기", "컴퓨터", "저장장치", "주변기기", "저장장치", "컴퓨터", "주변기기", "저장장치", "컴퓨터"],

    "product": ["키보드", "마우스", "노트북", "USB", "마우스", "USB", "노트북", "키보드", "SSD", "모니터"],

    "quantity": [2, 1, 1, 5, 3, 4, 1, 2, 1, 1],

    "unit_price": [30000, 15000, 1200000, 12000, 15000, 12000, 1200000, 30000, 90000, 250000]

}
```

```python
sns.set_theme(font='Malgun Gothic')
plt.figure(figsize=(10, 5))  # 

sns.lineplot(
      x=[1,2,3,4],
      y=[10,20,15,30],
      marker='o',
      markersize=10,
      linewidth=2.5,
      color='#1f77b4' #RGB  #붙이고 (RRGGBB) 특정 코드 모르겠으면 #00000으로 넣고 컬러블럭 클릭해서 바꿔라
)

plt.title('선 그래프',fontsize=16,fontweight='bold')
plt.xlabel('x')
plt.ylabel('y')

sns.despine()

plt.show()
```

#### 선 그래프

- 날짜별 매출계산

```python
df.groupby('order_date') ['total_price'].sum().reset_index()
```

```python
daily_sales = df.groupby('order_date') ['total_price'].sum().reset_index()
```

```python
sns.set_theme(font='Malgun Gothic')
plt.figure(figsize=(10, 5))

plt.plot(
    daily_sales['order_date'],
    daily_sales['total_price'],
    marker='o',
    markersize=10,
    linewidth=2.5,
    color="#E91212"
)
plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표시
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].dt.strftime('%Y-%m-%d')
)

plt.show()
```

##### 히스토그램

- 표준편차, 분포와 관계있는 그래프

- 숫자 데이터의 분포를 확인할때 사용

```python
plt.figure(figsize=(10,5))

plt.hist(df['total_price'],bins=5) #bins 구간갯수
plt.title('구매금액 분포도',fontsize =15)
plt.ticklabel_format(style='plain', axis='x')
plt.xlabel('구매금액 분포')
plt.ylabel('개수')

plt.show()
```

#### 산점도

- 두 숫자 데이터간의 관계를 확인 그래프
- 데이터 분석시 가장 많이 사용

```python
plt.figure(figsize=(10,5))

# 산점도
plt.scatter(x=df['quantity'],y=df['total_price'])
plt.title('구매수량,금액간의 관계도',fontsize =15)
plt.xlabel('구매갯수')
plt.ylabel('구매금액')

plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

##### 박스플롯

- 데이터의 이상치와 분포를 같이 확인할때 사용


```python
plt.figure(figsize=(10, 5))

plt.boxplot(df['total_price'])

plt.title('구매금액 분포와 이상치', fontsize=15)
plt.xticks([1])
plt.ylabel('구매금액')

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

##### 선 그래프 마커위치에 데이터 표시

```python
sns.set_theme(font='Malgun Gothic')
plt.figure(figsize=(10, 5))

plt.plot(
    daily_sales['order_date'],
    daily_sales['total_price'],
    marker='o',
    markersize=10,
    linewidth=2.5,
    color="#DA91918B")

#마커 위치에 실제 데이터 표시
for x,y in zip(daily_sales['order_date'],daily_sales['total_price']):
    plt.text(
        x,y,
        f"{y:,}",#y 값 =금액
        ha='center',# 가로축 정렬 left, center,right - 정렬 이름 반대
        va='bottom' # 세로축 top,center, bottom - 정렬 이름 반대
    )

plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표시
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].dt.strftime('%Y-%m-%d')
)

plt.show()
```

##### Seaborn으로 손쉽게 차트 구현

- Mapplotlib은 판다스DF를 바로 사용 못함
- Seaborn은 DF를 매개변수로 활용가능

```python
plt.figure(figsize=(10,5))

# 막대그래프 plt .bar ()-seaborn barplot으로 변경

sns.barplot(data=cate_sales,x='category',y='total_price')
plt.title('항목별 판매금액',fontsize=15,fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain',axis='y')
```

#### seaborn heatmap

- 숫자 데이터 분포를 색상으로 표현하는 그래프


```python
df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
corr = df[['quantity', 'unit_price', 'total_price']].corr()
plt.figure(figsize=(10, 5))
# 히트맵
sns.heatmap(data=corr, annot=True, cmap='Blues')
plt.title('구매 수량·단가·판매금액의 상관관계')
plt.show()
```
