### 조건 조회

- SQL WHERE 절과 동일

```python
df['score'] >=90
```

```python
df[df['score']>=90]
```

- numpy와 동일한 조건 연산
- 파이썬:and, or, not 사용
- numpy, pandas: & ,| , ~ 사용
- 연산자 우선 순위로 각 조건은 반드시 () 를 사용할 것

```python
#점수가 90점 초과이고 나이가 21살 이상인 학생 데이터

df[(df['score'] > 90) & (df['age'] >= 21)]
```

```python
#점수가 90점 초과이거나 나이가 21살 이상이거나 둘중에 하나로 참인 데이터 조회

df[(df['score'] > 90) | (df['age'] >= 21)]
```
