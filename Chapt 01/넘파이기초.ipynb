- 결측치를 정리방법: 결측치는 제거하거나 (건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기
 

#### 이상치 핸들링

- 일반적인 범위를 벗어나는 비정상 데이터
- 예로 부산 1월 일별 평균기온에 - 90도, 습도 99%

```python
temperatures = np.array([
    22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0
])

temperatures
```

```python
valid_temps =temperatures[
    (~np.isnan (temperatures)) &  #nan은 무조건 제거
    (temperatures >= 15.0) &     # 최저기온이 15아래로 내려갈수 없음
    (temperatures <= 50)    #최고 기온이 50 이상 오를 수 없음
]

valid_temps
```

- 전처리 (PreProcessing) 분석 이전에 이상치, 결측치를 없애는 작업

#### 매출 데이터 전처리

- 결측값 개수 출력
- 결측값 제외
- 0원 10만원 내의 정상 매출만 남김
- 정상 매출의 평균 최대값, 최소값 출력
- 정상 매출에서 평균 이상인 값만 출력 

```python
# 파이썬 리스트에서 none 대신 np.nan으로 결측치를 입력하면 numpy 패키지 연산이 가능

raw_sales =[12000, 15000, np.nan, 30000, -5000, 25000, 9999999,18000, np.nan,22000]

raw_sales
```

```python
#Numpy 배열로 바꿔줌
np_raw_sales = np.array(raw_sales)

np_raw_sales
```

```python
# 결측값 확인

np.isnan(np_raw_sales)
```

```python
np.isnan(raw_sales) #파이썬으로 실행하도 나옴
```

```python
#결측값 개수 확인

np.isnan(np_raw_sales).sum()
```

```python
# 결측치 제외

valid_sales = np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

valid_sales
```

```python
# 평균 매출액 20333.3원

mean_sales =valid_sales.mean()
mean_sales
```

```python
valid_sales.min()
```

```python
valid_sales.max()
```

```python
# 평균 매출액보다 높은 매출 찾기

valid_sales[valid_sales> mean_sales]
```
