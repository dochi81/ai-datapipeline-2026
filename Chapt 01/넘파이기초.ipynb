### Python 리스트 (배열)과 Numpy 배열비교

- 파이썬에 기존 리스트 (배열) 자료구조 존재, 속도적면에서 Numpy가 우세

### 파이썬으로 배열작업

```python
prices = [10000, 15000, 12000, 30000, 25000]  # 5개의 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(price * 0.9) # 10% 할인

discount_prices # 노트북 코드쉘에서 하나만 출력하려면 print() 생략가능
```

 ### 넘파이로 배열작업

```python
import numpy as np

prices = [10000,15000, 12000, 30000, 25000]

np_prices = np.array(prices) # 넘파이 배열로 변경
np_discount_prices = np_prices * 0.9

np_discount_prices.tolist()
```

-numpy 배열은 반복문을 사용하지 않아도 배열 전체에 계산을 수행할 수 있음. 벡터화 연산

#### Numpy 데이터타입

- int8 : -128 ~ 127 (총 256)
- int16 :-32768 ~ 32767 (총 65536)
- int32 : 32 비트정수
- int64 : 64 비트정수  (numpy 기본 정수)
- uin8 : 0 ~ 255
- uint16: 0 ~ 65536
- float 16 : 상수
- float 32 : 32 비트 상수
- float 64 : 64 비트 상수  (numpy 기본 상수)
- complex64 : 복소수
- complex128: 128 비트 복소수
- bool: true/ false
- str_: 문자열
- object_: 객체
